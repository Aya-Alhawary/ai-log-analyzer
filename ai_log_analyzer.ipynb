```python
#with open ("sample_logs.txt","w") as file:
#  file.write(logs)

#read file
with open("sample_logs.txt","r") as file:
  content = file.readlines()

#print(content)
#print(type(content))
counts = {}

for line in content:
  word = line.split(":")[0]

  if word == "INFO":
    if "INFO" not in counts:
      counts["INFO"] = 0
    counts["INFO"] += 1
  elif word == "ERROR":
    if "ERROR" not in counts:
      counts["ERROR"] = 0
    counts["ERROR"] += 1
  elif word == "WARNING":
    if "WARNING" not in counts:
      counts["WARNING"] = 0
    counts["WARNING"] += 1
  else:
    print(f"not supported")

print(counts)
with open("summary_rpt.txt","w") as file:
  for key,value in counts.items():
    file.write(f"{key}:{value}\n")
```
