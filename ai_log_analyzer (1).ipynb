```python
#Version 2

with open("sample_logs.txt","r") as file:
  content = file.readlines()

counts = {
    "INFO" : 0,
    "ERROR" : 0,
    "WARNING" : 0,
    "DEBUG" : 0,
    "CRITICAL": 0
}

for line in content:
  word = line.split(":")[0]

  if word in counts:
    counts[word] += 1

print(counts)

with open("summary_rpt.txt","w") as file:
  for key,value in counts.items():
    file.write(f"{key} : {value}\n")

with open("filtered.txt","w") as file:
  for line in content:
    if "ERROR" in line:
      file.write(line)
```
